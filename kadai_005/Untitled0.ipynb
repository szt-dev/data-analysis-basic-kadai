```python
# Step1. ライブラリのインポート
import matplotlib.pyplot as plt
import numpy as np

# Step2. 描画領域全体(Figure)と6個のサブプロット(Axes)を作成（2行×3列）
fig, axes = plt.subplots(2, 3, tight_layout=True)

# Step3. 各領域にグラフを描画

# (0,0) 折れ線グラフ
y0 = [1, 2, -5, 2]
axes[0, 0].plot(y0)
axes[0, 0].set_title("Line")

# (0,1) sin関数
x1 = np.linspace(0, 10, 100)
y1 = 2 + 2 * np.sin(2 * x1)
axes[0, 1].plot(x1, y1, linewidth=2.0)
axes[0, 1].set(xlim=(0, 10), xticks=np.arange(0, 10, 2),
               ylim=(-1, 5), yticks=np.arange(-1, 6))
axes[0, 1].set_title("sin")

# (0,2) ヒストグラム
np.random.seed(0)
x2 = np.random.randn(1000)
axes[0, 2].hist(x2, bins=30)
axes[0, 2].set_title("Histogram")

# (1,0) 散布図
x3 = np.random.rand(50)
y3 = np.random.rand(50)
axes[1, 0].scatter(x3, y3)
axes[1, 0].set_title("Scatter")

# (1,1), (1,2) は記載なし（空き領域）

# Step4. グラフを表示
plt.show()
```
